```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Python/loan_risk_prediction_dataset.csv")

print(df.head())
print(df.info())
```

```python
df = df[
    [
        'Age',
        'Income',
        'LoanAmount',
        'CreditScore',
        'YearsExperience',
        'Gender',
        'Education',
        'City',
        'EmploymentType',
        'LoanApproved'
    ]
]
```

```python

print(df.isnull().sum())
```

```python


df.dropna(inplace=True)

```

```python
display(df.describe())
```

```python
print("Average Loan Amount:",df['LoanAmount'].mean())
print("Average Income:",df['Income'].mean())
```

```python
print(df['LoanApproved'].value_counts())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='LoanApproved',data=df)
plt.show()
```

```python
sns.boxplot(
x='LoanApproved',
y='Income',
data=df
)
plt.show()
```

```python
sns.countplot(
x='EmploymentType',
hue='LoanApproved',
data=df
)
plt.xticks(rotation=90)
plt.show()
```

```python
sns.scatterplot(
x='Income',
y='LoanAmount',
data=df
)
plt.show()
```

```python
sns.heatmap(
df.corr(numeric_only=True),
annot=True
)
plt.show()
```
